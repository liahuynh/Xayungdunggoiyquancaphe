```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 1. Đọc dữ liệu sạch
df = pd.read_csv(r"C:\Duong\Dan\Toi\File\dkktdl_sach.csv")

# 2. Chọn các đặc trưng (features) số học để đưa vào thuật toán phân cụm
# Ở đây ta chọn: Giá trung bình, Điểm đánh giá, Độ yên tĩnh, và Khoảng cách đến HUIT
features = ['Gia_trung_binh', 'Diem_danh_gia(Rating)', 'Do_yen_tinh', 'Khoang_cach_HUIT_km']
X = df[features].copy()

# 3. Chuẩn hóa dữ liệu (StandardScaler rất quan trọng để các cột có thang đo đồng đều)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Xác định số cụm K (ví dụ chọn K = 3 nhóm quán) và chạy thuật toán K-Means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# 5. Xem thử kết quả phân cụm cho các quán đầu tiên
print(df[['Ten_quan', 'Gia_trung_binh', 'Do_yen_tinh', 'Cluster']].head(10))

# (Tùy chọn) In ra giá trị trung bình của từng cụm để đặt tên cho dễ hiểu
print("\n--- ĐẶC ĐIỂM TRUNG BÌNH CỦA TỪNG CỤM ---")
print(df.groupby('Cluster')[features].mean())
```
